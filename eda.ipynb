```python
df_clean = df.copy()

for col in df_clean.columns:
    df_clean[col] = (
        df_clean[col]
        .astype(str)
        .str.replace(",", "", regex=False)
        .str.replace("%", "", regex=False)
    )

    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
```

```python
num_cols = df_clean.select_dtypes(include="number").columns
num_cols
```

```python
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
```

```python
numeric_cols = [
    'Amount',
    'Value',
    'CountryCode',
    'PricingStrategy',
    'FraudResult'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df[numeric_cols].dtypes
```

```python
df[numeric_cols].describe().T
```

```python
summary = pd.DataFrame({
    'Mean': df[numeric_cols].mean(),
    'Median': df[numeric_cols].median(),
    'Std Dev': df[numeric_cols].std(),
    'Skewness': df[numeric_cols].skew(),
    'Kurtosis': df[numeric_cols].kurt()
})
```

```python
summary
```

```python
df[numeric_cols].hist(
    figsize=(14,10),
    bins=30,
    edgecolor='black'
)

plt.suptitle('Distribution of Numerical Features')
plt.tight_layout()
plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(6,4))

    sns.histplot(df[col], kde=True, bins=30)

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.show()
```

```python
cat_cols = [
    'CurrencyCode',
    'ProviderId',
    'ProductId',
    'ProductCategory',
    'ChannelId'
]
```

```python
for col in cat_cols:
    plt.figure(figsize=(10,5))

    sns.countplot(
        data=df,
        x=col,
        order=df[col].value_counts().index
    )

    plt.title(f'Distribution of {col}')
    plt.xticks(rotation=45)

    plt.show()
```

```python
corr_matrix = df[numeric_cols].corr()

corr_matrix
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()
```

```python
missing_values = df.isnull().sum()

missing_values[missing_values > 0]
```

```python
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percent
})

missing_df[missing_df['Missing Count'] > 0]
```

```python
plt.figure(figsize=(12,6))

sns.heatmap(df.isnull(), cbar=False)

plt.title('Missing Values Heatmap')
plt.show()
```

```python
for col in numeric_cols:
    plt.figure(figsize=(8,4))

    sns.boxplot(x=df[col])

    plt.title(f'Boxplot of {col}')

    plt.show()
```

```python
sns.countplot(data=df, x='FraudResult')

plt.title('Fraud Result Distribution')
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='FraudResult', y='Amount')

plt.title('Transaction Amount vs Fraud Result')
plt.show()
```

 The dataset contains both numerical and categorical variables related to transaction activity.
Transaction Amount and Value are highly right-skewed, indicating many small transactions and a few extremely large transactions.
Boxplots reveal significant outliers in transaction amounts.
Fraud cases appear imbalanced, with non-fraud transactions dominating the dataset.
Some numerical variables show moderate correlation, while others are weakly related.
