```python
import pandas as pd

# Example transaction data
df = pd.DataFrame({
    "customer_id": [1, 1, 1, 2, 2, 3],
    "transaction_amount": [100, 250, 150, 300, 200, 400],
    "transaction_date": pd.to_datetime([
        "2025-01-01", "2025-01-10", "2025-01-20",
        "2025-01-05", "2025-01-15", "2025-01-08"
    ])
})

# Aggregate features per customer
customer_features = (
    df.groupby("customer_id")
      .agg(
          total_transaction_amount=("transaction_amount", "sum"),
          average_transaction_amount=("transaction_amount", "mean"),
          max_transaction_amount=("transaction_amount", "max"),
          min_transaction_amount=("transaction_amount", "min"),
          transaction_count=("transaction_amount", "count")
      )
      .reset_index()
)

print(customer_features)
```

```python
customer_features = (
    df.groupby("customer_id")
      .agg(
          total_transaction_amount=("transaction_amount", "sum"),
          average_transaction_amount=("transaction_amount", "mean"),
          transaction_amount_std=("transaction_amount", "std"),
          transaction_count=("transaction_amount", "count")
      )
      .fillna(0)
      .reset_index()
)
```

```python
import pandas as pd

# Average transaction amount per customer
avg_transaction_amount = (
    df.groupby("customer_id")["transaction_amount"]
      .mean()
      .reset_index(name="average_transaction_amount")
)

print(avg_transaction_amount)
```

```python
customer_features = (
    df.groupby("customer_id")
      .agg(
          total_transaction_amount=("transaction_amount", "sum"),
          average_transaction_amount=("transaction_amount", "mean")
      )
      .reset_index()
)

print(customer_features)
```

```python
import pandas as pd

# Transaction count per customer
transaction_count = (
    df.groupby("customer_id")
      .size()
      .reset_index(name="transaction_count")
)

print(transaction_count)
```

```python
customer_features = (
    df.groupby("customer_id")
      .agg(
          transaction_count=("transaction_amount", "count")
      )
      .reset_index()
)

print(customer_features)
```

```python
import pandas as pd

# Standard deviation of transaction amounts per customer
transaction_std = (
    df.groupby("customer_id")["transaction_amount"]
      .std()
      .fillna(0)  # customers with only one transaction
      .reset_index(name="transaction_amount_std")
)

print(transaction_std)
```
